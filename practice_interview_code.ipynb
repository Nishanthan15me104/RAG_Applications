```python

# 1. Maximum Subarray Sum (The Classic Kadane's)
# Given an integer array, find the contiguous subarray (containing at least one number) which has the largest sum and return its sum.
nums = [-2, 1, -3, 4, -1, 2, 1, -5, 4]

def max_sum(nums : list ) -> int:
  curr_max = float('-inf')
  global_max = float('-inf')

  for num in nums:
    curr_max = max(curr_max + num , num)
    global_max = max(curr_max, global_max)
  return global_max
print(max_sum(nums=nums))


```

```python
# 2. Maximum Product Subarray
# Given an integer array, find the contiguous subarray within an array (containing at least one number) which has the largest product.

# Input: nums = [2, 3, -2, 4]
nums = [2, -2,-20]

def max_prod(nums: list ) ->  int :

  curr_max = nums[0]
  curr_min = nums[0]
  global_max = nums[0]
  for num in nums[1:]:
    # old_max = float('inf')
    # global_min = float('inf')
    print(f"--- Processing {num} ---")
    print(f"Start: max={curr_max}, min={curr_min}")
    old_min = curr_min
    old_max = curr_max
    curr_min = min(old_min * num , old_max * num, num)

    curr_max = max(old_max * num, old_min * num, num)

    # global_min = min(curr_min, global_min)
    global_max = max(curr_max, global_max)
    print(f"End:   max={curr_max}, min={curr_min}, global={global_max}")

  return global_max

print(max_prod(nums = nums))

```

```python
# 3. Maximum Circular Subarray Sum
# Given a circular array (where the last element is connected to the first), find the maximum sum of a non-empty subarray.

# Input: nums = [5, -3, 5]
nums = [5, -3, 5]

def rev_kandane(nums : list) -> int:
  curr_max =float('-inf')
  curr_min = float('inf')
  global_max = float('-inf')
  global_min = float('inf')
  final_max = float('-inf')
  total_sum = 0

  for num in nums :
    curr_max = max(num , curr_max + num)
    global_max = max(curr_max , global_max)

    total_sum += num

    curr_min = min(num , curr_min + num)
    global_min = min(curr_min , global_min)

  final_max = max(global_max, total_sum - global_min)

  return global_max if global_max < 0  else final_max

print(rev_kandane(nums = nums))

```

```python
# 2. Minimum Number of Refueling Stops
# Imagine a car with a limited gas tank.

# Input: target distance, start_fuel, and a list of stations where stations[i] = [distance_from_start, gas_available].

# The Task: You want to reach the target with the minimum number of stops.

# The "Greedy" Hook: Drive as far as you can. If you run out of fuel, look back at all the stations you passed and "virtually" stop at the one that had the most gas.

target = 100
start_fuel = 10
station_distances = [10, 20, 30, 60]
station_gas_amounts = [60, 30, 30, 40]

import heapq

def min_refuel( target , start_fuel , station_distances, station_gas_amounts):

  station = sorted(zip(station_distances,station_gas_amounts), key = lambda x : x[0], reverse = False) #[(10, 60), (20, 30), (30, 30), (60, 40)]


  station.append((target , 0))# append enf distace and 0 to calculate final destination
  heap = []
  covered_dis = 0
  fuel_left = start_fuel # holds currnt fuel
  refuel_count = 0
  # heap logic
  for dis , gas in station:
    curr_dis = dis - covered_dis
    while curr_dis > fuel_left:
      if not heap:
        return -1
      fuel_left += -(heapq.heappop(heap))
      refuel_count += 1
      # print (f" refuel_count : {refuel_count}")

    covered_dis += curr_dis
    fuel_left  -= curr_dis
    print(f"fueleft : {fuel_left} | coverd_dis: {covered_dis}")
    heapq.heappush(heap,-gas)


  return covered_dis ,fuel_left, refuel_count,heap


print(min_refuel(target =target, station_gas_amounts =station_gas_amounts , station_distances = station_distances, start_fuel = start_fuel))


```

```python
# 5. Task Scheduler
# Input: A list of tasks (e.g., ['A', 'A', 'A', 'B', 'B']) and a cooling time n.
# The Task: Same tasks cannot be executed within n intervals of each other. Find the minimum time to finish all tasks.
# The "Greedy" Hook: Always try to schedule the task that has the highest remaining count first. This "Greedy" choice prevents you from being stuck with a bunch of the same tasks at the very end with huge idle times.
import heapq
from collections import deque
tasks = ['A', 'A', 'A', 'B', 'B']
n = 2
def schedule_task(tasks : list , n : int) -> int :
  freq_dict ={} # {'A': 3, 'B': 2}

  for task in tasks:
    freq_dict[task] = freq_dict.get(task,0) + 1

  freq_list = [-count  for count in freq_dict.values() ] # [-3, -2]

  heapq.heapify(freq_list)#[-3, -2]

  queue = deque()
  time = 0
  task = 0
  while freq_list or queue:
    time += 1
    if freq_list:
      task = heapq.heappop(freq_list)
      task += 1
      if task != 0:
        queue.append((task,(time+n)))

    if  queue and time == queue[0][1]:
      pop = queue.popleft()
      heapq.heappush(freq_list, pop[0])


  return time

print(schedule_task(tasks = tasks , n= n))
```

```python
# 6. Reorganize String
# The Goal: Given a string s (e.g., "aaab"), rearrange the characters so that any two adjacent characters are not the same.
# If it’s impossible, return an empty string "".
string = "aaabbb"

import heapq

def reorganize_string(string : str )-> str:
  #frequecy
  freq = {} # {'a': 3, 'b': 1}
  for char in string:
    freq[char] = freq.get(char,0) + 1
  #greedy heap
  freq_list = sorted([(-count,char) for char , count in freq.items()],key = lambda x : x[0], reverse =True)
  # heapq.heapify
  heapq.heapify(freq_list)
  if -freq_list[0][0] > (len(string) + 1) // 2:
        return ""
  result_list = []
  # prev_count2 = 0
  # prev_char2 = ""
  prev_count1 = 0
  prev_char1 = ""
  while freq_list:
    if prev_count1 != 0 and prev_char1 != "":
      heapq.heappush(freq_list,(prev_count1,prev_char1))
    count , char = heapq.heappop(freq_list)
    result_list.append(char)
    count += 1
    # prev_count2 = prev_count1
    # prev_char2 = prev_char1
    prev_count1 = count
    prev_char1 = char


  return "".join(result_list)

print(reorganize_string(string = string))
```
